Load New York State Parks shapefile

```python
### load libraries

### folder/file management
import os
import pathlib

### tabular data
import pandas as pd

### handle geospatial vector data
import geopandas as gpd
```

```python
from pathlib import Path

print(Path.home())
```

```python
### create  data directory
data_dir = os.path.join(

    ### point to home directory
    pathlib.Path.home(),

    ### point to folders where your data will live
    'Earth_data',
    ### make a project directory for this assignment
    'final_project'
    )

os.makedirs(data_dir, exist_ok= True)
```

```python
### get url for New York State Parks shapefile
NYSParks_url = ("https://services.arcgis.com/1xFZPtKn1wKC6POA/ArcGIS/rest/services/"
              "NYS_Park_Polygons/FeatureServer/0/query?outFields=*&where=1%3D1&f=geojson")

### name the shapefile directory name and file name
NYSParks_dir = os.path.join(data_dir, "NYSParks")

### make the NYS Parks directory
os.makedirs(NYSParks_dir, exist_ok= True)

### join NYS Parks shapefile path
NYSParks_path = os.path.join(NYSParks_dir, "NYS_Park_Polygon.shp")

### download the data once
if not os.path.exists(NYSParks_path):

    ### open file at url
    NYSParks_gdf = gpd.read_file(NYSParks_url)

    ### save it to file path just defined
    NYSParks_gdf.to_file(NYSParks_path)

### open up the NYS Parks boundaries - make sure it will open outside of conditional statement above
NYSParks_gdf = gpd.read_file(NYSParks_path)
NYSParks_gdf

```

```python
# plot the NYS Parks quickly to check download
NYSParks_gdf.plot()
```

```python
### GBIF data download
### Collection Code: EBIRD
### Scientific Name 
### GADM ID is USA.33.24_1 = Brooklyn, NY
### Shirley Chisholm State Park's (SCSP) final public opening phase was in November, 2020.
### Using occurrence data from it's public opening until end of dataset Dec. 31, 2025

### point to folder where csv is saved
gbif_path = r"C:\Users\honey\Earth_data\final_project\savspa\0015718-261118162104854.csv"
gbif_path
```

```python
### Load the data
gbif_df = pd.read_csv(
    gbif_path, 
    delimiter = '\t', ### shows that data is tab delimited
    usecols = ['gbifID', 'decimalLatitude', 'decimalLongitude', 'eventDate', 'day', 
                 'month', 'year', 'dateIdentified', 'locality', 'individualCount']
)

### check out the data
gbif_df.head()
```

```python
gbif_df['individualCount'].max()
```

```python
gbif_df['year'].min()
```

```python
### convert the dataframe to a geodataframe
gbif_gdf = (
    gpd.GeoDataFrame(
        gbif_df, 

        ### set geometry for geodataframe from lat/lon of GBIF record
        geometry = gpd.points_from_xy(
            gbif_df.decimalLongitude, 
            gbif_df.decimalLatitude), 

        ### set the coordinate reference system
        crs="EPSG:4326")
    
    ### select the desired columns
    [['year', 'month', 'individualCount','geometry']] ### gbifID Will output no matter what because it is index
)

### check out the geodataframe
gbif_gdf
```
